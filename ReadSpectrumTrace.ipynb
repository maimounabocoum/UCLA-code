```python
import numpy as np
import matplotlib.pyplot as plt

# Full Path to your file
filename = r"D:\ucla data\spectrum\ArgonLS_63p3dBm_5900torr_500kHzdrive_25msIntTime_ST006871__1__19-46-08-601.txt"

# Read file
with open(filename, "r") as f:
    lines = f.readlines()

# Find beginning of spectral data
for i, line in enumerate(lines):
    if "Begin Spectral Data" in line:
        start_index = i + 1
        break

# Read numerical data
data = np.loadtxt(lines[start_index:])

# Separate columns
wavelength = data[:, 0]
intensity = data[:, 1]

# Plot spectrum
plt.figure(figsize=(10,5))
plt.plot(wavelength, intensity)

plt.xlabel("Wavelength (nm)")
plt.ylabel("Intensity")
plt.title("Spectrometer Spectrum")

plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------
# Full Path to your file
# ---------------------------------------------------

filename = r"D:\ucla data\spectrum\ArgonLS_63p3dBm_5900torr_500kHzdrive_25msIntTime_ST006871__1__19-46-08-601.txt"
filename = r"D:\ucla data\spectrum\ArgonLS_63p3dBm_5900torr_500kHzdrive_10msIntTime_ST006871__0__19-45-34-044.txt"
filename = r"D:\ucla data\spectrum\ArgonLS_60p1dBm_5900torr_400kHzdrive_8msIntTime_ST006871__0__19-29-05-825.txt"

# ---------------------------------------------------
# Read file
# ---------------------------------------------------

with open(filename, "r") as f:
    lines = f.readlines()

# Find beginning of spectral data
for i, line in enumerate(lines):
    if "Begin Spectral Data" in line:
        start_index = i + 1
        break

# Read numerical data
data = np.loadtxt(lines[start_index:])

# Separate columns
wavelength = data[:, 0]
intensity = data[:, 1]

# ---------------------------------------------------
# Lines to mark
# ---------------------------------------------------

markers = [
    (385.60, "385.6 Si II"),
    (447.1, "447.1 N2"),
    (470.9, "470.9 N2 / N2+"),
    (480.6, "480.6 Ar I"),
    (515.14, "515.1 C I"),
    (577.21, "577.2 Si I"),
    (590.60, "590.6 N2"),
    (603.3797,"603.4 Ar I"),
    (621.73, "621.7 Ar I"),
    (636.38, "636.3 O I"),
    (656.3, "656.3 Hα"),
    (667.73, "667.7 N2"),
    (687.3,"687.3 Ar I"),
    (696.5, "696.5 Ar I"),
    (711.5, "711.5 C I"),
    (727.3, "727.3 Ar I"),
    (738.6014,"738.6 Ar I"),
    (750.5935,"750.6 Ar I"),
    (763.5, "763.5 Ar I"),
    (772.5887,"772.5 Ar I"),
    (777.19, "777.2 O I"),
    (795.0362,"795.0 Ar I")
    

]

# ---------------------------------------------------
# Plot spectrum
# ---------------------------------------------------

fig, ax = plt.subplots(figsize=(10,5))

ax.plot(wavelength, intensity)

# Add vertical lines + labels
for wl, label in markers:

    # Vertical line
    ax.axvline(
        wl,
        linestyle="--",
        alpha=0.7
    )

    # Label
    ax.text(
        wl,
        ax.get_ylim()[1] * 0.95,
        label,
        rotation=90,
        verticalalignment='top',
        horizontalalignment='right',
        fontsize=9
    )

# ---------------------------------------------------
# Labels
# ---------------------------------------------------

ax.set_xlabel("Wavelength (nm)")
ax.set_ylabel("Intensity")
ax.set_title("Spectrometer Spectrum")

ax.grid(True)

plt.tight_layout()
plt.show()
```
